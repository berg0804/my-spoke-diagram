```python
import pandas as pd
import networkx as nx
nodes_df = pd.read_csv("nodes.csv")
edges_df = pd.read_csv("edges.csv")
```

```python
G = nx.Graph()  # or nx.DiGraph() for directed relationships

# Add nodes with attributes
for _, row in nodes_df.iterrows():
    G.add_node(row["ID"], label=row["Label"])

# Add edges
for _, row in edges_df.iterrows():
    G.add_edge(row["Source"], row["Target"])

# ID to label
id_to_label = dict(zip(nodes_df["ID"], nodes_df["Label"]))

# Label to ID
label_to_id = {v: k for k, v in id_to_label.items()}

label1 = "Marcel Dionne"
label2 = "Max Domi"

# Convert to node IDs
node1 = label_to_id[label1]
node2 = label_to_id[label2]

# Find shortest path (in IDs)
path_ids = nx.shortest_path(G, source=node1, target=node2)

# Convert back to labels
path_labels = [id_to_label[n] for n in path_ids]

print(" → ".join(path_labels))

```

```python
# Example: Starting with a label
selected_label = "No Trade"
selected_id = label_to_id[selected_label]

# Get connected nodes (neighbors)
neighbors = list(G.neighbors(selected_id))

# Create a list of label + degree info
connections = []
for neighbor_id in neighbors:
    label = id_to_label.get(neighbor_id, str(neighbor_id))
    degree = G.degree[neighbor_id]
    connections.append((label, degree))

# Sort by degree (optional)
connections = sorted(connections, key=lambda x: -x[1])

# Display
print(f"Connections for {selected_label} (degree: {G.degree[selected_id]}):\n")
for label, degree in connections:
    print(f"- {label} (degree: {degree})")
```

```python
# 📚 Import Libraries
import pandas as pd
import networkx as nx

# -------------------------------
# 📥 Load CSV Data
# -------------------------------
# Load nodes and edges
nodes_df = pd.read_csv("nodes.csv")
edges_df = pd.read_csv("edges.csv")
nodes_df["Label"] = nodes_df["Label"].astype(str).str.strip()

# Build the graph
G = nx.Graph()
for _, row in nodes_df.iterrows():
    G.add_node(row["ID"], label=row["Label"])
for _, row in edges_df.iterrows():
    G.add_edge(row["Source"], row["Target"])

print(f"✅ Graph loaded: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")

# -------------------------------
# 🔎 Extract Connections Function
# -------------------------------
def get_connections_as_dataframe(graph, search_name, exclude_words=None, max_degree=5):
    """
    Find all connections for a given node, excluding paths through unwanted words.
    """
    search_name_clean = str(search_name).strip().lower()

    # Build filtered graph that excludes unwanted nodes
    graph_copy = graph.copy()
    if exclude_words:
        nodes_to_remove = [
            n for n, d in graph_copy.nodes(data=True)
            if any(word.lower() in str(d.get("label", "")).lower() for word in exclude_words)
        ]
        graph_copy.remove_nodes_from(nodes_to_remove)

    # Find node ID for search_name in filtered graph
    node_id = None
    for n, data in graph_copy.nodes(data=True):
        label = str(data.get("label", "")).strip()
        if label.lower() == search_name_clean:
            node_id = n
            break

    if node_id is None:
        raise ValueError(f"❌ '{search_name}' not found in graph after applying exclusions.")

    # Get shortest paths from the node (respects excluded nodes)
    all_paths = nx.single_source_shortest_path(graph_copy, node_id, cutoff=max_degree)

    seen_pairs = set()
    rows = []

    for target_node, path in all_paths.items():
        if target_node == node_id:
            continue  # Skip self

        degree = len(path) - 1

        for parent, child in zip(path[:-1], path[1:]):
            parent_label = str(graph_copy.nodes[parent].get("label", str(parent)))
            child_label = str(graph_copy.nodes[child].get("label", str(child)))
            pair = (parent_label, child_label)
            if pair not in seen_pairs:
                rows.append({
                    "Parent": parent_label,
                    "Child": child_label,
                    "Degree": degree
                })
                seen_pairs.add(pair)

    return pd.DataFrame(rows)

# -------------------------------
# 🔎 Run Search and Export
# -------------------------------
# Set your search parameters here
search_name = "Rem Pitlick"
exclude_words = []
max_degree = 20

try:
    df_connections = get_connections_as_dataframe(G, search_name, exclude_words, max_degree)

    if df_connections.empty:
        print(f"⚠️ No connections found for '{search_name}' after applying exclusions.")
    else:
        print(f"✅ Found {len(df_connections)} connections for '{search_name}'")
        display(df_connections.head(10))  # Show preview of first 10 rows

        # Export to CSV
        filename = f"connections_{search_name.replace(' ', '_').lower()}.csv"
        df_connections.to_csv(filename, index=False)
        print(f"📁 Exported to: {filename}")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
# 📚 Import Libraries
import pandas as pd
import networkx as nx

# -------------------------------
# 📥 Load CSV Data
# -------------------------------
nodes_df = pd.read_csv("nodes.csv")
edges_df = pd.read_csv("edges.csv")
nodes_df["Label"] = nodes_df["Label"].astype(str).str.strip()

# Build the graph
G = nx.Graph()
for _, row in nodes_df.iterrows():
    G.add_node(row["ID"], label=row["Label"])
for _, row in edges_df.iterrows():
    G.add_edge(row["Source"], row["Target"])

print(f"✅ Graph loaded: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")
```

```python
# -------------------------------
# 🔎 Function to Find Path
# -------------------------------
def find_path_between_nodes(graph, start_name, end_name, exclude_words=None):
    """
    Finds the shortest path between two nodes, excluding unwanted nodes.
    """
    start_name_clean = str(start_name).strip().lower()
    end_name_clean = str(end_name).strip().lower()

    # Filter graph to exclude unwanted nodes
    graph_copy = graph.copy()
    if exclude_words:
        nodes_to_remove = [
            n for n, d in graph_copy.nodes(data=True)
            if any(word.lower() in str(d.get("label", "")).lower() for word in exclude_words)
        ]
        graph_copy.remove_nodes_from(nodes_to_remove)

    # Find IDs for start and end nodes
    start_id = None
    end_id = None
    for n, data in graph_copy.nodes(data=True):
        label = str(data.get("label", "")).strip().lower()
        if label == start_name_clean:
            start_id = n
        if label == end_name_clean:
            end_id = n

    if start_id is None:
        raise ValueError(f"❌ Start node '{start_name}' not found in graph.")
    if end_id is None:
        raise ValueError(f"❌ End node '{end_name}' not found in graph.")

    # Find shortest path
    try:
        path = nx.shortest_path(graph_copy, source=start_id, target=end_id)
    except nx.NetworkXNoPath:
        raise ValueError(f"❌ No path found between '{start_name}' and '{end_name}'.")

    # Build DataFrame of parent-child relationships
    rows = []
    for parent, child in zip(path[:-1], path[1:]):
        parent_label = str(graph_copy.nodes[parent].get("label", str(parent)))
        child_label = str(graph_copy.nodes[child].get("label", str(child)))
        rows.append({"Parent": parent_label, "Child": child_label})

    return pd.DataFrame(rows)
```

```python
def get_connections_as_dataframe(graph, search_name, exclude_words=None, max_degree=5):
    """
    Find all connections for a given node, excluding paths through unwanted words.
    Returns both IDs and labels.
    """
    search_name_clean = str(search_name).strip().lower()

    # Build filtered graph that excludes unwanted nodes
    graph_copy = graph.copy()
    if exclude_words:
        nodes_to_remove = [
            n for n, d in graph_copy.nodes(data=True)
            if any(word.lower() in str(d.get("label", "")).lower() for word in exclude_words)
        ]
        graph_copy.remove_nodes_from(nodes_to_remove)

    # Find node ID for search_name in filtered graph
    node_id = None
    for n, data in graph_copy.nodes(data=True):
        label = str(data.get("label", "")).strip()
        if label.lower() == search_name_clean:
            node_id = n
            break

    if node_id is None:
        raise ValueError(f"❌ '{search_name}' not found in graph after applying exclusions.")

    # Get shortest paths from the node (respects excluded nodes)
    all_paths = nx.single_source_shortest_path(graph_copy, node_id, cutoff=max_degree)

    seen_pairs = set()
    rows = []

    for target_node, path in all_paths.items():
        if target_node == node_id:
            continue  # Skip self

        degree = len(path) - 1

        for parent, child in zip(path[:-1], path[1:]):
            parent_label = str(graph_copy.nodes[parent].get("label", str(parent)))
            child_label = str(graph_copy.nodes[child].get("label", str(child)))

            pair = (parent, child)  # use IDs for uniqueness
            if pair not in seen_pairs:
                rows.append({
                    "Parent_ID": parent,
                    "Parent_Label": parent_label,
                    "Child_ID": child,
                    "Child_Label": child_label,
                    "Degree": degree
                })
                seen_pairs.add(pair)

    return pd.DataFrame(rows)
```

```python
import pandas as pd

# Load CSVs
nodes = pd.read_csv("nodes.csv")
edges = pd.read_csv("edges.csv")

# Normalize IDs → int → str
nodes["ID"] = nodes["ID"].astype(float).astype(int).astype(str)
edges["Source"] = edges["Source"].astype(float).astype(int).astype(str)
edges["Target"] = edges["Target"].astype(float).astype(int).astype(str)

# Build lookup dictionary {id: label}
id_to_label = dict(zip(nodes["ID"], nodes["Label"]))

# Replace IDs in edges with labels
edges["source_label"] = edges["Source"].map(id_to_label)
edges["target_label"] = edges["Target"].map(id_to_label)
edges.drop(['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4',
       'Unnamed: 5', 'Unnamed: 6'], axis=1, inplace=True)
edges = edges.sort_values(by='target_label')



# Save output
edges.to_csv("edges_with_labels.csv", index=False)

# Debug: check any still missing
missing_targets = edges.loc[edges["target_label"].isna(), "Target"].unique()
print("⚠️ Missing target IDs:", missing_targets)
print("✅ edges_with_labels.csv saved!")
```

```python
import pandas as pd
import json

# Load your edge data
df = pd.read_csv("edge3.csv")

# Build a mapping of trade → list of players
trade_to_players = {}
for _, row in df.iterrows():
    trade = row["source"].strip()
    player = row["target"].strip()
    trade_to_players.setdefault(trade, []).append(player)

# Build a player-to-player graph (no trade nodes)
player_graph = {}
for trade, players in trade_to_players.items():
    for i, p1 in enumerate(players):
        for j, p2 in enumerate(players):
            if i != j:
                player_graph.setdefault(p1, []).append(p2)

# Remove duplicates
for player, neighbors in player_graph.items():
    player_graph[player] = sorted(set(neighbors))

# Save clean player-player JSON
with open("player_graph.json", "w") as f:
    json.dump(player_graph, f, indent=2)

print("✅ Cleaned player_graph.json created successfully!")
```
